### Activity 13: Identifying Duplicate Entries in Clothing Stock

Create a program that reads the `clothing_stock.txt` file and identifies if there are any duplicate entries (both item and size). Use a set to find and print these duplicates.

**Steps:**
1. Read the file and store each item in a set.
2. Keep track of duplicate entries.
3. Print any duplicates found.
---



`````python
# Step 1: Read the file and check for duplicates
seen_items = set()
duplicates = set()

with open("clothing_stock.txt", "r", encoding="utf-8") as file:
    for line in file:
        item = line.strip()  # Remove whitespace/newline

        if item in seen_items:
            duplicates.add(item)
        else:
            seen_items.add(item)

# Step 2: Print duplicates found
if duplicates:
    print("Duplicate entries found:")
    for item in duplicates:
        print(item)
else:
    print("No duplicate entries found.")
`````

1. Read the file and extract both clothing items and sizes.

`````python
# Dictionary to store clothing items and their sizes
item_sizes = {}

# Step 1: Read the file and extract items and sizes
with open("clothing_stock.txt", "r", encoding="utf-8") as file:
    for line in file:
        # Example format: T-Shirt,M
        item, size = line.strip().split(",")

        # Step 2: Store sizes in a set to keep only unique sizes
        if item not in item_sizes:
            item_sizes[item] = set()

        item_sizes[item].add(size)

# Step 3: Count different sizes for each item
size_counts = {}

for item, sizes in item_sizes.items():
    size_counts[item] = len(sizes)

# Print the results
print("Number of different sizes available for each clothing item:")

for item, count in size_counts.items():
    print(f"{item}: {count} size(s)")
`````

### Activity 15: Managing Study Groups

You are responsible for managing a study group for a course. Each student can be part of multiple study groups, and you need to track which groups a student is enrolled in. Students can join or leave study groups. What data structure would you choose to manage this efficiently, and why?

**Example data:**

```python
{"John": ["Math", "Physics"], "Emily": ["Physics", "Biology"], "Alex": ["Math"]}

Steps:

1. Choose a data structure that allows quick access to the groups each student belongs to.
2. Write a program that lets students join and leave study groups.
3. Implement a function to display all groups a student is part of.
---



`````python
# Dictionary: student -> set of study groups
study_groups = {
    "John": {"Math", "Physics"},
    "Emily": {"Physics", "Biology"},
    "Alex": {"Math"}
}

# Function to join a study group
def join_group(student, group):
    if student not in study_groups:
        study_groups[student] = set()

    study_groups[student].add(group)
    print(f"{student} joined {group}")

# Function to leave a study group
def leave_group(student, group):
    if student in study_groups and group in study_groups[student]:
        study_groups[student].remove(group)
        print(f"{student} left {group}")
    else:
        print(f"{student} is not in {group}")

# Function to display a student's groups
def show_groups(student):
    if student in study_groups:
        print(f"{student}'s groups: {', '.join(study_groups[student])}")
    else:
        print(f"No groups found for {student}")

# Example usage
join_group("John", "Chemistry")
leave_group("Emily", "Physics")

show_groups("John")
show_groups("Emily")

print("\nAll Study Groups:")
for student, groups in study_groups.items():
    print(f"{student}: {groups}")
`````

### Activity 17: Counting Words in Les_Brown.txt

Read the content from `Les_Brown.txt`, and then count the number of occurrences of each word in the text. Write the word count results to a new file called `word_count.txt`.

**Steps:**
1. Open and read the contents of `Les_Brown.txt`.
2. Split the content into words and count the frequency of each word.
3. Write the word count in the format: `word: count` to a new file called `word_count.txt`.

---



`````python
# Step 1: Open and read the contents of Les_Brown.txt
with open("Les_Brown.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Convert text to lowercase for case-insensitive counting
text = text.lower()

# Remove common punctuation
for char in ".,!?;:\"'()":
    text = text.replace(char, "")

# Step 2: Split into words and count frequencies
words = text.split()

word_counts = {}

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# Step 3: Write the word counts to word_count.txt
with open("word_count.txt", "w", encoding="utf-8") as file:
    for word, count in sorted(word_counts.items()):
        file.write(f"{word}: {count}\n")

print("Word counts have been written to 'word_count.txt'.")
`````

### Activity 20: Managing Event Tickets

You are developing a system to manage event tickets for concerts or sports events. Each event has a limited number of tickets available, and users can purchase tickets if they are still available. You need to track which users have purchased tickets for specific events and how many are still available. What data structure would you use for this task, and why?

**Example data:**
- Available tickets: `{"Concert": 100, "Football Game": 50, "Theater Show": 75}`
- Purchasers: `{"Concert": ["John", "Emily"], "Football Game": ["Alex"], "Theater Show": ["Sam", "John"]}`

**Steps:**
1. Choose a data structure to track ticket availability and purchasers efficiently.
2. Write a program that handles ticket purchases and updates the list of purchasers.
3. Implement functions to check availability and view who has purchased tickets.
---



`````python
# Available tickets for each event
available_tickets = {
    "Concert": 100,
    "Football Game": 50,
    "Theater Show": 75
}

# Purchasers for each event
purchasers = {
    "Concert": ["John", "Emily"],
    "Football Game": ["Alex"],
    "Theater Show": ["Sam", "John"]
}

# Function to purchase a ticket
def purchase_ticket(event, user):
    if event not in available_tickets:
        print("Event not found.")
        return

    if available_tickets[event] > 0:
        available_tickets[event] -= 1

        if event not in purchasers:
            purchasers[event] = []

        purchasers[event].append(user)

        print(f"{user} purchased a ticket for {event}.")
    else:
        print(f"Sorry, {event} is sold out.")

# Function to check ticket availability
def check_availability(event):
    if event in available_tickets:
        print(f"{event}: {available_tickets[event]} tickets remaining")
    else:
        print("Event not found.")

# Function to view purchasers
def view_purchasers(event):
    if event in purchasers:
        print(f"Purchasers for {event}:")
        for person in purchasers[event]:
            print("-", person)
    else:
        print("No purchasers found for this event.")

# Example usage
purchase_ticket("Concert", "Diana")

check_availability("Concert")

view_purchasers("Concert")
`````

### Activity 21: Personal Finance App

You are creating a personal finance app that allows users to categorize their expenses and track spending by category (e.g., food, entertainment, travel). Users can add expenses, remove expenses, and see a summary of their spending. What data structure would you choose to store and organize the expenses, and why?

**Example data:**

```python
{"Food": [25, 40, 15], "Entertainment": [50, 20], "Travel": [300, 150]}
```
Steps:

1. Choose a data structure that lets users easily add and track expenses by category.
2. Write a program that allows users to input expenses, remove them, and view their total spending by category.
3. Implement a function to display a summary of all spending.
---


`````python
# Dictionary to store expenses by category
expenses = {
    "Food": [25, 40, 15],
    "Entertainment": [50, 20],
    "Travel": [300, 150]
}

# Add an expense
def add_expense(category, amount):
    if category not in expenses:
        expenses[category] = []

    expenses[category].append(amount)
    print(f"Added £{amount} to {category}")

# Remove an expense
def remove_expense(category, amount):
    if category in expenses and amount in expenses[category]:
        expenses[category].remove(amount)
        print(f"Removed £{amount} from {category}")
    else:
        print("Expense not found.")

# View total spending for a category
def view_category_total(category):
    if category in expenses:
        total = sum(expenses[category])
        print(f"{category} total: £{total}")
    else:
        print("Category not found.")

# Display a summary of all spending
def display_summary():
    print("\nExpense Summary")
    print("-" * 20)

    overall_total = 0

    for category, amounts in expenses.items():
        total = sum(amounts)
        overall_total += total
        print(f"{category}: £{total}")

    print("-" * 20)
    print(f"Overall Total: £{overall_total}")

# Example usage
add_expense("Food", 30)
remove_expense("Entertainment", 20)

view_category_total("Food")

display_summary()
`````
