### Task 7: Simple Sentiment Analyser

Write a function `analyse_sentiment(reviews)` that:
1. Cleans each review (`strip()`, `lower()`)
2. Classifies it as **positive**, **negative**, or **neutral** using keyword lists
3. Returns a dictionary with counts and a list of which reviews are negative

**Methods/Functions to use:** `strip()`, `lower()`, `any()`, `in`


```python
reviews = [
    "  AMAZING product, Love it!!!  ",
    "terrible experience   ",
    "  It was okay  ",
    "WORST purchase ever",
    "  fantastic quality  ",
    "poor customer service",
    "  Not bad, not great  "
]

def analyse_sentiment(reviews):

    positive_words = ["amazing", "love", "fantastic", "great"]
    negative_words = ["terrible", "worst", "poor", "bad"]

    result = {
        "positive": 0,
        "negative": 0,
        "neutral": 0,
        "negative_reviews": []
    }

    for review in reviews:

        clean_review = review.strip().lower()

        if any(word in clean_review for word in positive_words):
            result["positive"] += 1

        elif any(word in clean_review for word in negative_words):
            result["negative"] += 1
            result["negative_reviews"].append(clean_review)

        else:
            result["neutral"] += 1

    return result


reviews = [
    "  AMAZING product, Love it!!!  ",
    "terrible experience   ",
    "  It was okay  ",
    "WORST purchase ever",
    "  fantastic quality  ",
    "poor customer service",
    "  Not bad, not great  "
]

print(analyse_sentiment(reviews))

```

### Task 8: Data Validation Pipeline

Write a function `validate_record(record)` that takes a CSV string and checks:
1. It has exactly 4 fields (name, age, email, phone)
2. Name contains only letters and spaces
3. Age is a valid number between 18 and 100
4. Email contains `@` and `.`
5. Phone starts with `+` or `0`

Return a list of **error messages** (empty list = valid record).

Use `all()` to check if ALL fields are non-empty first.

**Methods to use:** `split()`, `isdigit()`, `isalpha()`, `startswith()`, `all()`


records = [
    "Alice Smith,25,alice@email.com,+447123456789",
    "Bob,abc,bob@email,07123456789",
    "Charlie123,30,charlie@mail.com,447123456789",
    ",22,missing@mail.com,+440000000000"
]


```python
def validate_record(record):
    errors = []

    # Split record into fields
    fields = [field.strip() for field in record.split(",")]

    # Check number of fields
    if len(fields) != 4:
        errors.append("Record must contain exactly 4 fields.")
        return errors

    name, age, email, phone = fields

    # Check all fields are non-empty
    if not all(fields):
        errors.append("All fields must be non-empty.")

    # Name validation
    if not all(char.isalpha() or char.isspace() for char in name):
        errors.append("Name must contain only letters and spaces.")

    # Age validation
    if not age.isdigit():
        errors.append("Age must be a valid number.")
    else:
        age_num = int(age)
        if age_num < 18 or age_num > 100:
            errors.append("Age must be between 18 and 100.")

    # Email validation
    if "@" not in email or "." not in email:
        errors.append("Email must contain '@' and '.'.")

    # Phone validation
    if not (phone.startswith("+") or phone.startswith("0")):
        errors.append("Phone must start with '+' or '0'.")

    return errors


records = [
    "Alice Smith,25,alice@email.com,+447123456789",
    "Bob,abc,bob@email,07123456789",
    "Charlie123,30,charlie@mail.com,447123456789",
    ",22,missing@mail.com,+440000000000"
]

for record in records:
    result = validate_record(record)

    print(f"\nRecord: {record}")

    if not result:
        print("Valid record")
    else:
        print("Errors:")
        for error in result:
            print("-", error)
```

### Task 9: Word Frequency Counter

Write a function `word_frequency(text)` that:
1. Converts text to lowercase
2. Removes punctuation (`.`, `,`, `!`, `?`)
3. Splits into words
4. Returns a **dictionary** of word → count
5. Print the top 5 most common words

**Methods to use:** `lower()`, `replace()`, `split()`, `count()`


```python
text = """Data science is the study of data. Data scientists use data to find patterns.
Machine learning is a part of data science. Scientists study data every day!
The science of data is growing, and data is everywhere."""

def word_frequency(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.replace(".", "")
    text = text.replace(",", "")
    text = text.replace("!", "")
    text = text.replace("?", "")

    # Split into words
    words = text.split()

    # Count words
    frequency = {}

    for word in words:
        frequency[word] = words.count(word)

    # Print top 5 common words
    print("Top 5 words:")

    top_words = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

    for word, count in top_words[:5]:
        print(word, "=", count)

    return frequency


text = """
Python is great. Python is easy!
Coding in Python is fun, and Python is powerful.
"""

print(word_frequency(text))
    
   
```

### Task 10: Password Strength Checker

Write a function `check_password(password)` that scores a password (0–5):

| Rule | Points |
|------|--------|
| At least 8 characters | +1 |
| Contains uppercase letter | +1 |
| Contains lowercase letter | +1 |
| Contains a digit | +1 |
| Contains a special character (`@#$%!`) | +1 |

Print the score and a label: **Weak** (0-2), **Medium** (3-4), **Strong** (5).

If the password contains a space, print  and reject it.

**Methods to use:** `len()`, `isupper()`, `islower()`, `isdigit()`, `any()`


```python
test_passwords = [
    "hello",
    "Hello123",
    "My Pass!",
    "Str0ng@Pass",
    "12345678"
]

def check_password(password):
    # Reject passwords with spaces
    if " " in password:
        print(f"{password}: Invalid password (contains spaces)")
        return

    score = 0

    # At least 8 characters
    if len(password) >= 8:
        score += 1

    # Contains uppercase letter
    if any(char.isupper() for char in password):
        score += 1

    # Contains lowercase letter
    if any(char.islower() for char in password):
        score += 1

    # Contains a digit
    if any(char.isdigit() for char in password):
        score += 1

    # Contains a special character
    if any(char in "@#$%!" for char in password):
        score += 1

    # Determine label
    if score <= 2:
        label = "Weak"
    elif score <= 4:
        label = "Medium"
    else:
        label = "Strong"

    print(f"{password}: Score = {score}/5, {label}")


test_passwords = [
    "hello",
    "Hello123",
    "My Pass!",
    "Str0ng@Pass",
    "12345678"
]

for password in test_passwords:
    check_password(password)
```

### Task 11: Caesar Cipher Encoder & Decoder

Write **two** functions:
1. `encode(text, key)` — shifts each letter forward by `key` positions
2. `decode(text, key)` — reverses the shift to get the original text

Rules:
- Only shift letters (a-z). Leave spaces, digits, and punctuation unchanged
- Wrap around: `z` shifted by 1 → `a`
- Preserve case: uppercase stays uppercase
- Verify: `decode(encode(text, key), key)` should return the original text

**Methods to use:** `ord()`, `chr()`, `isalpha()`, `isupper()`, `islower()`


```python

def encode(text, key):
    result = []
    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result.append(chr((ord(ch) - base + key) % 26 + base))
        else:
            result.append(ch)
    return ''.join(result)


def decode(text, key):
    return encode(text, -key)


# Test
original = "Hello World! 123"
key = 5

encoded = encode(original, key)
decoded = decode(encoded, key)

print(f"Original: {original}")
print(f"Encoded:  {encoded}")
print(f"Decoded:  {decoded}")
print(f"Round-trip success: {decoded == original}")
```

### Task 12: Text Data Cleaning Pipeline

Build a `clean_dataset(records)` function that processes raw text records. Each record is a string with fields separated by `|`.

The function should:
1. Skip records where **any** field is empty
2. Clean names to **Title Case** and remove extra spaces
3. Validate that age is a number (18–120)
4. Validate email format (contains `@` and has a `.` after the `@`)
5. Standardise country names to **uppercase**
6. Return two lists: `valid_records` and `invalid_records` with reasons

**Methods to use:** `split()`, `strip()`, `title()`, `upper()`, `isdigit()`, `all()`, `find()`


```python
raw_data = [
    "  john SMITH  | 25 | john.smith@email.com | united kingdom ",
    "  MARY  johnson | abc | mary@email | france",
    " | 30 | test@test.com | germany",
    "  bob wilson  | 17 | bob@mail.com | spain  ",
    "  alice Brown  | 28 | alice.b@data.org | italy  ",
    "charlie DAVIS | 35 | | usa"
]

def clean_dataset(records):
    valid_records = []
    invalid_records = []

    for record in records:
        # Split and clean fields
        fields = [field.strip() for field in record.split("|")]

        if len(fields) != 4:
            invalid_records.append((record, "Record must contain 4 fields"))
            continue

        name, age, email, country = fields

        # Check for empty fields
        if not all(fields):
            invalid_records.append((record, "One or more fields are empty"))
            continue

        # Clean name to Title Case
        name = " ".join(name.split()).title()

        # Validate age
        if not age.isdigit():
            invalid_records.append((record, "Age must be a number"))
            continue

        age = int(age)

        if age < 18 or age > 120:
            invalid_records.append((record, "Age must be between 18 and 120"))
            continue

        # Validate email
        at_pos = email.find("@")

        if at_pos == -1:
            invalid_records.append((record, "Email missing '@'"))
            continue

        if email.find(".", at_pos) == -1:
            invalid_records.append((record, "Email must contain '.' after '@'"))
            continue

        # Standardise country
        country = country.upper()

        # Store cleaned valid record
        valid_records.append({
            "name": name,
            "age": age,
            "email": email,
            "country": country
        })

    return valid_records, invalid_records


valid_records, invalid_records = clean_dataset(raw_data)

print("VALID RECORDS")
for record in valid_records:
    print(record)

print("\nINVALID RECORDS")
for record, reason in invalid_records:
    print(f"{reason}: {record}")
```

### Task 13: Analyse & Report on Text Corpus

Write a function `text_report(text)` that produces a complete analysis:

1. **Basic stats:** Total characters, words, sentences, average word length
2. **Character breakdown:** Count of uppercase, lowercase, digits, spaces, punctuation
3. **Top 5 words:** Most frequently used words (exclude common words: 'the', 'a', 'is', 'and', 'of', 'to', 'in')
4. **Longest and shortest words**
5. **Print a formatted report** using string formatting

**Methods to use:** `split()`, `lower()`, `replace()`, `count()`, `isupper()`, `islower()`, `isdigit()`, `isspace()`, `join()`, `center()`, `ljust()`


```python
corpus = """Artificial Intelligence is transforming every industry. From healthcare to finance,
AI systems are making decisions that affect millions of people. Machine Learning,
a subset of AI, enables computers to learn from data without being explicitly programmed.
Deep Learning takes this further with neural networks inspired by the human brain.
In 2024, Large Language Models like GPT have shown remarkable abilities in understanding
and generating human language. The future of AI is both exciting and challenging."""


def text_report(text):
    # -----------------------------
    # Basic Statistics
    # -----------------------------
    total_characters = len(text)

    words = text.split()
    total_words = len(words)

    total_sentences = text.count(".") + text.count("!") + text.count("?")

    # Clean words for length calculations
    clean_text = text.lower()
    for char in ".,!?":
        clean_text = clean_text.replace(char, "")

    clean_words = clean_text.split()

    avg_word_length = (
        sum(len(word) for word in clean_words) / len(clean_words)
        if clean_words else 0
    )

    # -----------------------------
    # Character Breakdown
    # -----------------------------
    uppercase = sum(1 for ch in text if ch.isupper())
    lowercase = sum(1 for ch in text if ch.islower())
    digits = sum(1 for ch in text if ch.isdigit())
    spaces = sum(1 for ch in text if ch.isspace())

    punctuation_chars = ".,!?;:'\"()-"
    punctuation = sum(1 for ch in text if ch in punctuation_chars)

    # -----------------------------
    # Word Frequency
    # -----------------------------
    stop_words = {"the", "a", "is", "and", "of", "to", "in"}

    frequencies = {}

    for word in clean_words:
        if word not in stop_words:
            frequencies[word] = frequencies.get(word, 0) + 1

    top_5 = sorted(
        frequencies.items(),
        key=lambda item: item[1],
        reverse=True
    )[:5]

    # -----------------------------
    # Longest and Shortest Words
    # -----------------------------
    unique_words = list(set(clean_words))

    longest_word = max(unique_words, key=len)
    shortest_word = min(unique_words, key=len)

    # -----------------------------
    # Formatted Report
    # -----------------------------
    print("TEXT ANALYSIS REPORT".center(50, "="))
    print()

    print("Basic Statistics".ljust(25, "-"))
    print(f"Total Characters : {total_characters}")
    print(f"Total Words      : {total_words}")
    print(f"Total Sentences  : {total_sentences}")
    print(f"Average Word Len : {avg_word_length:.2f}")
    print()

    print("Character Breakdown".ljust(25, "-"))
    print(f"Uppercase Letters: {uppercase}")
    print(f"Lowercase Letters: {lowercase}")
    print(f"Digits           : {digits}")
    print(f"Spaces           : {spaces}")
    print(f"Punctuation      : {punctuation}")
    print()

    print("Top 5 Words".ljust(25, "-"))
    for word, count in top_5:
        print(f"{word.ljust(15)} {count}")

    print()

    print("Word Length Analysis".ljust(25, "-"))
    print(f"Longest Word     : {longest_word}")
    print(f"Shortest Word    : {shortest_word}")

    print()
    print("=" * 50)


# Run the report
text_report(corpus)

```

### Task 14: Build a Simple Search Engine

Create a function `search(documents, query)` that:
1. Takes a list of document strings and a search query
2. Returns documents **ranked** by how many times the query words appear
3. The search should be **case-insensitive**
4. If query has multiple words, count matches for **each** word
5. Show a relevance score and highlight which query words were found

**Methods to use:** `lower()`, `split()`, `count()`, `any()`, `all()`, `strip()`, `replace()`


```python
documents = [
    "Python is a great programming language for data science and machine learning",
    "Data analysis with Python involves cleaning and processing large datasets",
    "Machine learning models can predict outcomes from historical data",
    "Web development with JavaScript and HTML is different from data work",
    "Python data science libraries include NumPy and Pandas for data processing"
]


def search(documents, query):
    query_words = query.lower().strip().split()

    results = []

    for document in documents:
        doc_lower = document.lower()

        # Remove simple punctuation if present
        for char in ".,!?":
            doc_lower = doc_lower.replace(char, "")

        words = doc_lower.split()

        score = 0
        found_words = []

        for query_word in query_words:
            matches = words.count(query_word)

            if matches > 0:
                score += matches
                found_words.append(query_word)

        # Only include documents containing at least one query word
        if any(word in words for word in query_words):
            results.append({
                "document": document,
                "score": score,
                "found_words": found_words,
                "all_words_found": all(
                    word in words for word in query_words
                )
            })

    # Sort by relevance score (highest first)
    results.sort(key=lambda item: item["score"], reverse=True)

    return results


# Example searches
query = "python data"

results = search(documents, query)

print(f"Search Query: '{query}'")
print("-" * 70)

for rank, result in enumerate(results, start=1):
    print(f"Rank #{rank}")
    print(f"Score: {result['score']}")
    print(f"Found Words: {', '.join(result['found_words'])}")

    if result["all_words_found"]:
        print("Match Type: ALL query words found")
    else:
        print("Match Type: Partial match")

    print(f"Document: {result['document']}")
    print("-" * 70)

```

Write a function `classify_logs(logs)` that takes a list of log strings and:

5. Phone starts with `+` or `0`

Return a list of **error messages** (empty list = valid record).

Use `all()` to check if ALL fields are non-empty first.

**Methods to use:** `split()`, `isdigit()`, `isalpha()`, `startswith()`, `all()`


```python
records = [
    "Alice Smith,25,alice@email.com,+447123456789",
    "Bob,abc,bob@email,07123456789",
    "Charlie123,30,charlie@mail.com,447123456789",
    ",22,missing@mail.com,+440000000000"
]

def validate_record(record):
    errors = []

    # Split the CSV record
    fields = record.split(",")

    # Check number of fields
    if len(fields) != 4:
        errors.append("Record must contain exactly 4 fields.")
        return errors

    name, age, email, phone = fields

    # Check all fields are non-empty
    if not all(field.strip() for field in fields):
        errors.append("All fields must be non-empty.")

    # Validate name (letters and spaces only)
    if not all(char.isalpha() or char.isspace() for char in name):
        errors.append("Name must contain only letters and spaces.")

    # Validate age
    if not age.isdigit():
        errors.append("Age must be a valid number.")
    else:
        age_num = int(age)
        if age_num < 18 or age_num > 100:
            errors.append("Age must be between 18 and 100.")

    # Validate email
    if "@" not in email or "." not in email:
        errors.append("Email must contain '@' and '.'.")

    # Validate phone
    if not (phone.startswith("+") or phone.startswith("0")):
        errors.append("Phone must start with '+' or '0'.")

    return errors


# Test the records
for record in records:
    result = validate_record(record)

    print(f"\nRecord: {record}")

    if not result:
        print("Valid record")
    else:
        print("Errors:")
        for error in result:
            print("-", error)

```

### Task 9: Word Frequency Counter

Write a function `word_frequency(text)` that:
1. Converts text to lowercase
2. Removes punctuation (`.`, `,`, `!`, `?`)
3. Splits into words
4. Returns a **dictionary** of word → count
5. Print the top 5 most common words

**Methods to use:** `lower()`, `replace()`, `split()`, `count()`


```python
text = """Data science is the study of data. Data scientists use data to find patterns.
Machine learning is a part of data science. Scientists study data every day!
The science of data is growing, and data is everywhere."""


def word_frequency(text):
    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    for char in ".,!?":
        text = text.replace(char, "")

    # Split into words
    words = text.split()

    # Count frequencies
    frequencies = {}

    for word in words:
        frequencies[word] = words.count(word)

    return frequencies


# Get word frequencies
freq = word_frequency(text)

print("Word Frequencies:")
for word, count in freq.items():
    print(f"{word}: {count}")

# Print top 5 most common words
print("\nTop 5 Most Common Words:")

top_5 = sorted(freq.items(), key=lambda item: item[1], reverse=True)[:5]

for word, count in top_5:
    print(f"{word}: {count}")

```

### Task 10: Password Strength Checker

Write a function `check_password(password)` that scores a password (0–5):

| Rule | Points |
|------|--------|
| At least 8 characters | +1 |
| Contains uppercase letter | +1 |
| Contains lowercase letter | +1 |
| Contains a digit | +1 |
| Contains a special character (`@#$%!`) | +1 |

Print the score and a label: **Weak** (0-2), **Medium** (3-4), **Strong** (5).

If the password contains a space, print  and reject it.

**Methods to use:** `len()`, `isupper()`, `islower()`, `isdigit()`, `any()`


```python
test_passwords = [
    "hello",
    "Hello123",
    "My Pass!",
    "Str0ng@Pass",
    "12345678"
]

def check_password(password):
    # Reject passwords containing spaces
    if " " in password:
        print(f"'{password}' -> Invalid password: contains spaces")
        return

    score = 0

    # Rule 1: At least 8 characters
    if len(password) >= 8:
        score += 1

    # Rule 2: Contains uppercase letter
    if any(char.isupper() for char in password):
        score += 1

    # Rule 3: Contains lowercase letter
    if any(char.islower() for char in password):
        score += 1

    # Rule 4: Contains a digit
    if any(char.isdigit() for char in password):
        score += 1

    # Rule 5: Contains a special character
    if any(char in "@#$%!" for char in password):
        score += 1

    # Determine strength label
    if score <= 2:
        label = "Weak"
    elif score <= 4:
        label = "Medium"
    else:
        label = "Strong"

    print(f"'{password}' -> Score: {score}/5 ({label})")


# Test all passwords
for pwd in test_passwords:
    check_password(pwd)

```

### Task 12: Text Data Cleaning Pipeline

Build a `clean_dataset(records)` function that processes raw text records. Each record is a string with fields separated by `|`.

The function should:
1. Skip records where **any** field is empty
2. Clean names to **Title Case** and remove extra spaces
3. Validate that age is a number (18–120)
4. Validate email format (contains `@` and has a `.` after the `@`)
5. Standardise country names to **uppercase**
6. Return two lists: `valid_records` and `invalid_records` with reasons

**Methods to use:** `split()`, `strip()`, `title()`, `upper()`, `isdigit()`, `all()`, `find()`


```python
raw_data = [
    "  john SMITH  | 25 | john.smith@email.com | united kingdom ",
    "  MARY  johnson | abc | mary@email | france",
    " | 30 | test@test.com | germany",
    "  bob wilson  | 17 | bob@mail.com | spain  ",
    "  alice Brown  | 28 | alice.b@data.org | italy  ",
    "charlie DAVIS | 35 | | usa"
]


import re

def clean_dataset(records):
    valid_records = []
    invalid_records = []

    for record in records:
        fields = record.split("|")

        # Rule 1: Skip records where any field is empty
        if len(fields) != 4 or any(f.strip() == "" for f in fields):
            invalid_records.append({
                "raw": record.strip(),
                "reasons": ["One or more fields are empty"]
            })
            continue

        name, age, email, country = [f.strip() for f in fields]
        reasons = []

        # Rule 2: Clean name to Title Case and remove extra spaces
        name = " ".join(name.split()).title()

        # Rule 3: Validate age is a number between 18–120
        if not age.isdigit() or not (18 <= int(age) <= 120):
            reasons.append(f"Invalid age '{age}' (must be a number between 18–120)")

        # Rule 4: Validate email format
        at_index = email.find("@")
        if at_index == -1 or "." not in email[at_index:]:
            reasons.append(f"Invalid email '{email}' (must contain '@' and a '.' after it)")

        # Rule 5: Standardise country to uppercase
        country = country.upper()

        if reasons:
            invalid_records.append({"raw": record.strip(), "reasons": reasons})
        else:
            valid_records.append({
                "name":    name,
                "age":     int(age),
                "email":   email,
                "country": country
            })

    return valid_records, invalid_records


# ── Test ───────────────────────────────────────────────────────────────────────
raw_data = [
    "  john SMITH  | 25 | john.smith@email.com | united kingdom ",
    "  MARY  johnson | abc | mary@email | france",
    " | 30 | test@test.com | germany",
    "  bob wilson  | 17 | bob@mail.com | spain  ",
    "  alice Brown  | 28 | alice.b@data.org | italy  ",
    "charlie DAVIS | 35 | | usa"
]

valid_records, invalid_records = clean_dataset(raw_data)

print("=" * 50)
print(f"  VALID RECORDS ({len(valid_records)})")
print("=" * 50)
for r in valid_records:
    print(f"  {r}")

print()
print("=" * 50)
print(f"  INVALID RECORDS ({len(invalid_records)})")
print("=" * 50)
for r in invalid_records:
    print(f"  Raw:     {r['raw']}")
    for reason in r['reasons']:
        print(f"  Reason:  {reason}")
    print()




```

### Task 13: Analyse & Report on Text Corpus

Write a function `text_report(text)` that produces a complete analysis:

1. **Basic stats:** Total characters, words, sentences, average word length
2. **Character breakdown:** Count of uppercase, lowercase, digits, spaces, punctuation
3. **Top 5 words:** Most frequently used words (exclude common words: 'the', 'a', 'is', 'and', 'of', 'to', 'in')
4. **Longest and shortest words**
5. **Print a formatted report** using string formatting

**Methods to use:** `split()`, `lower()`, `replace()`, `count()`, `isupper()`, `islower()`, `isdigit()`, `isspace()`, `join()`, `center()`, `ljust()`


```python
corpus = """Artificial Intelligence is transforming every industry. From healthcare to finance,
AI systems are making decisions that affect millions of people. Machine Learning,
a subset of AI, enables computers to learn from data without being explicitly programmed.
Deep Learning takes this further with neural networks inspired by the human brain.
In 2024, Large Language Models like GPT have shown remarkable abilities in understanding
and generating human language. The future of AI is both exciting and challenging."""


import string
from collections import Counter


def text_report(text):
    # -----------------------------
    # Basic Stats
    # -----------------------------
    total_characters = len(text)

    words = text.split()
    total_words = len(words)

    sentences = text.count('.') + text.count('!') + text.count('?')

    clean_words = [
        word.strip(string.punctuation)
        for word in words
        if word.strip(string.punctuation)
    ]

    avg_word_length = (
        sum(len(word) for word in clean_words) / len(clean_words)
        if clean_words else 0
    )

    # -----------------------------
    # Character Breakdown
    # -----------------------------
    uppercase = sum(1 for char in text if char.isupper())
    lowercase = sum(1 for char in text if char.islower())
    digits = sum(1 for char in text if char.isdigit())
    spaces = sum(1 for char in text if char.isspace())
    punctuation = sum(1 for char in text if char in string.punctuation)

    # -----------------------------
    # Top 5 Words
    # -----------------------------
    stop_words = {'the', 'a', 'is', 'and', 'of', 'to', 'in'}

    filtered_words = [
        word.strip(string.punctuation).lower()
        for word in words
        if word.strip(string.punctuation).lower() not in stop_words
    ]

    word_counts = Counter(filtered_words)
    top_5_words = word_counts.most_common(5)

    # -----------------------------
    # Longest & Shortest Words
    # -----------------------------
    unique_words = [
        word.strip(string.punctuation)
        for word in words
        if word.strip(string.punctuation)
    ]

    longest_word = max(unique_words, key=len)
    shortest_word = min(unique_words, key=len)

    # -----------------------------
    # Formatted Report
    # -----------------------------
    print("=" * 50)
    print("TEXT ANALYSIS REPORT")
    print("=" * 50)

    print("\nBASIC STATISTICS")
    print(f"Total Characters     : {total_characters}")
    print(f"Total Words          : {total_words}")
    print(f"Total Sentences      : {sentences}")
    print(f"Average Word Length  : {avg_word_length:.2f}")

    print("\nCHARACTER BREAKDOWN")
    print(f"Uppercase Letters    : {uppercase}")
    print(f"Lowercase Letters    : {lowercase}")
    print(f"Digits               : {digits}")
    print(f"Spaces               : {spaces}")
    print(f"Punctuation Marks    : {punctuation}")

    print("\nTOP 5 WORDS")
    for word, count in top_5_words:
        print(f"{word:<15} {count}")

    print("\nWORD LENGTH ANALYSIS")
    print(f"Longest Word         : {longest_word}")
    print(f"Shortest Word        : {shortest_word}")

    print("=" * 50)

corpus = """Artificial Intelligence is transforming every industry. From healthcare to finance,
AI systems are making decisions that affect millions of people. Machine Learning,
a subset of AI, enables computers to learn from data without being explicitly programmed.
Deep Learning takes this further with neural networks inspired by the human brain.
In 2024, Large Language Models like GPT have shown remarkable abilities in understanding
and generating human language. The future of AI is both exciting and challenging."""

text_report(corpus)

```

### Task 14: Build a Simple Search Engine

Create a function `search(documents, query)` that:
1. Takes a list of document strings and a search query
2. Returns documents **ranked** by how many times the query words appear
3. The search should be **case-insensitive**
4. If query has multiple words, count matches for **each** word
5. Show a relevance score and highlight which query words were found

**Methods to use:** `lower()`, `split()`, `count()`, `any()`, `all()`, `strip()`, `replace()`


```python
documents = [
    "Python is a great programming language for data science and machine learning",
    "Data analysis with Python involves cleaning and processing large datasets",
    "Machine learning models can predict outcomes from historical data",
    "Web development with JavaScript and HTML is different from data work",
    "Python data science libraries include NumPy and Pandas for data processing"
]

def search(documents, query):
    query_words = query.lower().split()
    results = []

    for doc in documents:
        doc_lower = doc.lower()

        score = 0
        found_words = []

        for word in query_words:
            count = doc_lower.count(word)

            if count > 0:
                score += count
                found_words.append(f"{word} ({count})")

        if score > 0:
            results.append((score, found_words, doc))

    # Sort by relevance score (highest first)
    results.sort(reverse=True, key=lambda x: x[0])

    print(f"\nSearch Query: '{query}'")
    print("-" * 70)

    if not results:
        print("No matching documents found.")
        return

    for rank, (score, found_words, doc) in enumerate(results, start=1):
        print(f"Rank #{rank}")
        print(f"Relevance Score: {score}")
        print(f"Found Words: {', '.join(found_words)}")
        print(f"Document: {doc}")
        print("-" * 70)


documents = [
    "Python is a great programming language for data science and machine learning",
    "Data analysis with Python involves cleaning and processing large datasets",
    "Machine learning models can predict outcomes from historical data",
    "Web development with JavaScript and HTML is different from data work",
    "Python data science libraries include NumPy and Pandas for data processing"
]

search(documents, "python data learning")
```
